```python
import os

os.environ["XLA_PYTHON_CLIENT_PREALLOCATE"] = "false"

import jax
import jax.numpy as jnp
import matscipy.neighbours
import numpy as onp

from run_benchmark import build_neighborlists, new_build_neighborlists
```

```python
CUTOFF = 3.0
PBC = (False,) * 3
```

```python
structures = onp.load("reference_results/nonperiodic/structures.npz")
n_structures = structures["positions"].shape[0]
n_particles = structures["positions"].shape[1]
```

```python
neighborlists = build_neighborlists(
    structures["positions"], structures["cells"], CUTOFF, PBC
)
```

```python
neighborlists_new = new_build_neighborlists(
    structures["positions"], structures["cells"], CUTOFF, PBC
)
```

```python
jnp.array_equal(neighborlists, neighborlists_new)
```

```python
for nbl, nbl_new in zip(neighborlists, neighborlists_new):
    assert (nbl[0] == nbl_new[0]).all()
    assert (nbl[1] == nbl_new[1]).all()
```
